```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Q1. Create a countplot to show the number of passengers by gender

```python
sns.countplot(data=df, x='Sex')
plt.title('Number of passengers by gender')
plt.xlabel('Gender')
plt.ylabel('Number of passenger')
plt.show()
```

Q2. Create a bar chart to show the number of passengers in each passenger class using differ

```python
sns.barplot(data=df, x='Pclass', y='PassengerId')
plt.title('Number of passengers in each passenger class')
plt.xlabel('Passenger class')
plt.ylabel('Number of passenger')
plt.show()
```

Q3. Create a histogram to show the distribution of passenger age

```python
sns.histplot(data=df, x='Age')
plt.title('Distribution of passenger age')
plt.xlabel('Age')
plt.ylabel('PaasengerID')
plt.show()
```

Q4. Create a boxplot to detect outliers in the fare column

```python
sns.boxplot(data=df, x='Fare')
plt.title('Outliers in the fare column')
plt.xlabel('Fare')
plt.ylabel('PassengerID')
plt.show()
```

Q5. Create a scatterplot to

a. show the relationship between age and fare.

b. analyze the relationship between family size (SibSp + Parch) and fare.

```python
sns.scatterplot(data=df, x='Age', y='Fare')
plt.title('Relationship between age and fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
df["FamilySize"] = df["SibSp"] + df["Parch"]
sns.scatterplot(data=df, x='FamilySize', y='Fare')
plt.title('Relationship between family size (SibSp + Parch) and fare')
plt.xlabel('size (SibSp + Parch)')
plt.ylabel('Fare')
plt.show()
```

Q6. Create a countplot to show survival count by gender

```python
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title('Survival count by gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passenger')
plt.show()
```

Q7. Create a bar chart to show survival by passenger class

```python
sns.barplot(data=df, x='Pclass', hue='Survived')
plt.title('servival by passenger class')
plt.xlabel('Passenger class')
plt.ylabel('Survived')
plt.show()
```

Q8. Create a heatmap to show correlation between numerical variables.


```python
corr=df.select_dtypes(include='number').corr()
plt.figure(figsize=(5,6))
sns.heatmap(corr, cmap='coolwarm')
plt.title('Correlation between numerical variables')
plt.show()
```
